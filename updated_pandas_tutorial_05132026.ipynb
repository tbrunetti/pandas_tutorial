## Generate new pandas objects
Two main data structures in pandas:  
* Series = 1-dimensional iteratable such as a list of integers, a dictionary of strings, an array of objects, etc...
* DataFrame - 2-dimensional that works like a table with rows and columns

### Series examples

```python
# generate a series
int_series_data = pd.Series([1, 5, 5, np.nan, 6, 8])
int_series_data
```

```python
# generate string series
str_series_data = pd.Series(["this", "is", "a", "string", "series", "it", "can", "contain", "duplicates", "duplicates", "count", "the", "dups", "duplicates"])
str_series_data
```

```python
# generate string series
# index equates to row names
str_series_data = pd.Series(["this", "is", "a", "string", "series", "it", "can", "contain", "duplicates", "duplicates", "count", "the", "dups", "duplicates"], 
                            index=["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m", "n"])
str_series_data
```

```python
# generate string series
# index equates to row names - DO NOT NEED TO BE UNIQUE!
str_series_data = pd.Series(["this", "is", "a", "string", "series", "it", "can", "contain", "duplicates", "duplicates", "count", "the", "dups", "duplicates"], 
                            index=["a", "a", "a", "a", "e", "f", "g", "h", "c", "j", "k", "l", "b", "c"])
str_series_data
```

### Dataframe example

```python
# generate dataframes 
# date_range() produces a numpy array
dates_df = pd.date_range(start = "20130101", 
                         periods=6, 
                         freq="D",     
                         inclusive="both")
dates_df
```

```python
# generate data frame by dictionary
df2_nonamed_index = pd.DataFrame(data= {
     "A": 1.0,
     "B": pd.Timestamp("20130102"),
     "C": pd.Series(1, index = range(8), dtype = "float32"),
     "D": np.array([3] * 8, dtype = "int32"),
     "E": pd.Categorical(values=["test", "train", "test", "train", "test", "train", "test", "train"]),
     "F": "foo"
     }
)
df2_nonamed_index
```

```python
# generate data frame by dictionary
# now add index (row names)
df2 = pd.DataFrame(data= {
     "A": 1.0,
     "B": pd.Timestamp("20130102"),
     "C": pd.Series(1, index = range(4), dtype = "float32"),
     "D": np.array([3] * 4, dtype = "int32"),
     "E": pd.Categorical(values=["test", "train", "test", "train"]),
     "F": "foo"
     },index=['sample {}'.format(num) for num in range(1,5)]
)
df2
```

```python
# generate data frame by dictionary
# now add index (row names) - DOES NOT NEED TO BE UNIQUE!!
df2 = pd.DataFrame(data= {
     "A": 1.0,
     "B": pd.Timestamp("20130102"),
     "C": pd.Series(1, index = range(7), dtype = "float32"),
     "D": np.array([3] * 7, dtype = "int32"),
     "E": pd.Categorical(values=["test", "train", "unknown", "train", "train", "test", "test"]),
     "F": "foo",
     "G": np.random.randint(low = -5, high = 7, size = 7),
     "H": np.random.ranf(size = 7),
     "I": [True, False, False, False, False, True, True]
     },
     index=["sample 1", "sample 2", "sample 1", "sample 2", "sample 3", "sample 4", "sample 5"]
)
df2
```

```python
# print column "A"
df2.A
```

```python
# axis = 0 (rows), axis = 1 (columns)
# count the number of non-NaN entries in each row
# in pandas, NaN, NaT, None, pandas.NA are all considered as NA
df2.count(axis = 0) 

# the results below would indicate that for column "C" all rows(index in our case samples) are all NaN/NA/NaT/None and that column "A", has 7 rows(samples in our case) values that are not NA
```

```python
# axis = 0 (rows), axis = 1 (columns)
# count the number of non-NaN, non-NaT, non-None, non-NA entries in each column
# in pandas, NaN, NaT, None, pandas.NA are all considered as NA
df2.count(axis = 1, numeric_only = False)

# the results below would indicate that 7 columns in the first occurence of "sample 1" row/index (sample in our case) all contain values that are not NA
# as a matter of fact, all samples (rows) have 7 columns that don't contain NA.  This is true because we have 8 columns, and we know column "C" is NA for all rows
```

## Selecting and Subsetting Data  
* Getitem (`[]`) - good for rearranging order of columns in a dataframe
* slicing (`:`) - good for selecting rows/indices
* by label (`.loc[]` or `.at[]`)
* by position (`.iloc[]` or `.iat[]`)
* boolean indexing (`>`, `<`, `>=`, `<=`, `==`, `&`, `|`, `.isin()`, `.query()`)
* keep original shape but show where (`.where()`) or mask values (`.mask()`)
* setting/updating

### Example: Getitem (`[]`) - good for rearranging order of columns in a dataframe

```python
# This will return a series for what is listed in column E
# df2["E"] and df2.E are equivalent, the main difference is the .E version can only be used if your columns only contain letter, numbers and underscores
# if you have spaces or special characters then the GetItem (["E"]) approach as shown below must be used
df2["E"]
```

```python
# select columns G, A, and B and rearrange them in the same order provided
df2[["G", "A", "B"]]
```

### Example: slicing (`:`) - good for selecting rows/indices


```python
# select the first two rows of the data frame
# note, the last number is non-inclusive and as with python, dataframes are 0-indexed
df2[0:2]
```

```python
# select the last 3 rows
df2[df2.shape[0]-3:df2.shape[0]]
```

### Example: by label (`.loc[]` or `.at[]`)


```python
# subset to all rows of columns E and G only
df2.loc[:, ["E", "G"]]
```

```python
# subset to rows called "sample 1" and of columns E and G only
df2.loc["sample 1", ["E", "G"]]
```

```python
# subset to rows called "sample 1" and "sample 2" and of columns E and G only
df2.loc[["sample 1", "sample 2"], ["E", "G"]]
```

```python
# For getting fast access to a scalar (equivalent to the prior method)
# but for single scalar values this is fastest
df2.at["sample 5", "G"]
```

### Example: by position (`.iloc[]` or `.iat[]`)


```python
# select first 2 rows and 4th row, but keep all columns
df2.iloc[[0,1,3],:]
```

```python
# keep all rows but select first 3 columns
df2.iloc[:,0:3]
```

```python
# get value of cell at index 1 (2nd row) and column index 7 (8th column)
df2.iloc[1,7]
```

```python
# get value of cell at index 1 (2nd row) and column index 7 (8th column)
# faster access for singular scalar value
df2.iat[1,7]
```

### boolean indexing (`>`, `<`, `>=`, `<=`, `==`, `&`, `|`, `.isin()`, `.query()`)

```python
# select all rows where G is negative
df2[df2['G'] < 0]
```

```python
# select all rows where G is negative and H is greater than or equal to 0.50
df2[(df2['G'] < 0) & (df2['H'] >= 0.50)] 
```

```python
# select all rows where E is equal to "train"
df2[(df2['E'] == "train")] 
```

```python
# select all rows where E is not equal to "train"
df2[(df2['E'] != "train")] 
```

```python
# select all rows where I is False (~ negates the default of True )
df2.loc[~df2['I']]
```

```python
# select all rows where I is True 
df2.loc[df2['I']]
```

```python
# select all rows where E is equal to "test"
# or where G is positive
df2[(df2['E'] == "test") | (df2["G"] > 0)] 
```

```python
# select all rows where E is equal to "test"
# or where G is positive
# equivalent to the above command; just a way to write expressions in shorthand
df2.query('(E == "test") | (G > 0)')
```

```python
# select rows where column E contains either test or train as categories
df2[df2['E'].isin(["test", "train"])]
```

```python
# mask all values in the original data where the statement below is not True
# i.e show me full row entries where column A is less than 15, all other rows
# should be NaN/NaT/None depending on series dtype for the column
df2.where(df2['A'] < 15)
```

```python
# inverse behavior of where
# i.e. show me full row values where this is false/not true - hence mask these rows
df2.mask(df2['A'] < 15)
```

### Example: setting/updating

```python
# make a new series
# Setting a new column automatically aligns the data by the indexes
# matches the index listed to the df index just like a merge by row name
pheno = pd.Series(["wt", "wt", "wt" ,"wt", "ko", "ko", "ko", "ko"],  [0, 2, 4, 6, 1, 3, 5, 7])
df2_nonamed_index["phenotype"] = pheno
df2_nonamed_index
```

```python
# change/set/update value of cell by label
df2.at["sample 4", "C"]  = 3
df2
```

```python
# change/set/update value of cell by position
# 1st row and 4th column (col index 3 b/c 0-indexed) set value to 5
df2.iat[0, 3]  = 5
df2
```

```python
# multiple each cell in column A for every row by 2 
# and overwrite the results
df2.loc[:,"A"] = df2.loc[:,"A"]*2
df2
```

```python
# of rows where column H is greater than 0.5, 
# set column A of those rows to 10
df2.loc[df2['H'] > 0.5, "A"] = 10
```
